```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib osx
from matplotlib import cm
from matplotlib.animation import FFMpegWriter
```

```python
L = 30
n = 50
X = np.linspace(0, L, n + 1)
Y = np.linspace(0, L, n + 1)
X, Y = np.meshgrid(X, Y)

T = np.zeros((n+1, n+1))
for i in range(10, 13):
    T[10:40, i] = 100

for i in range(13, 27):
    T[10:13, i] = 100
    T[23:26, i] = 100
    T[37:40, i] = 100
plt.imshow(T, origin='lower')
plt.show()
```

```python
fig = plt.figure(dpi=200)
metadata = dict(title='2D Heat Equation', artist='Henry Jugan',
                comment='Using the letter E for diffusion')
writer = FFMpegWriter(fps=15, metadata=metadata)

T_new = np.copy(T)
counter = 0
n_iter = 2000
frame_every = 10

with writer.saving(fig, "hw7_part1.mp4", dpi=200):
    while True:
        # Draw first, so frame 0 is the original E
        if counter % frame_every == 0:
            fig.clear()
            ax = fig.add_subplot(projection='3d')
            ax.plot_surface(X, Y, T, cmap=cm.coolwarm, vmin=0, vmax=100,
                            antialiased=False)
            ax.set_zlim(0, 100)
            ax.set_title(f"iteration {counter}")
            plt.draw()
            plt.pause(0.01)
            writer.grab_frame()

        if counter == n_iter:
            break

        for i in range(1, n):
            for j in range(1, n):
                T_new[i,j] = 1/4*(T[i+1,j] + T[i-1,j] + T[i,j+1] + T[i,j-1])
        T[:,:] = T_new[:,:]
        counter += 1
```

```python
fig = plt.figure(dpi=200)
metadata = dict(title='2D Heat Equation', artist='Henry Jugan',
                comment='Using the letter E for diffusion')
writer = FFMpegWriter(fps=15, metadata=metadata)

T_new = np.copy(T)
counter = 0
n_iter = 2000
frame_every = 10

with writer.saving(fig, "hw7_part2.mp4", dpi=200):
    while True:
        # Draw first, so frame 0 is the original E
        if counter % frame_every == 0:
            fig.clear()
            ax = fig.add_subplot(projection='3d')
            ax.plot_surface(X, Y, T, cmap=cm.coolwarm, vmin=0, vmax=100,
                            antialiased=False)
            ax.set_zlim(0, 100)
            ax.set_title(f"iteration {counter}    total heat = {T.sum():.0f}")
            plt.draw()
            plt.pause(0.01)
            writer.grab_frame()

        if counter == n_iter:
            break

        for i in range(1, n):
            for j in range(1, n):
                T_new[i,j] = 1/4*(T[i+1,j] + T[i-1,j] + T[i,j+1] + T[i,j-1])
        for i in range(n):
            for j in range(n):
                T_new[i,0] = ( T[i,0] + T[i+1,0] + T[i-1,0] + T[i,1] )/4.0
                T_new[i,50] = ( T[i,50] + T[i+1,50] + T[i-1,50] + T[i,1] )/4.0
                T_new[50,j] = ( T[50,j] + T[50, j+1] + T[50, j-1] + T[1, j])/4.0
                T_new[0,j] = ( T[0,j] + T[0, j+1] + T[0, j-1] + T[1, j])/4.0
        T_new[0,0] = ( 2.0*T[0,0] + T[1,0] + T[0,1] )/4.0
        T_new[50,0] = ( 2.0*T[0,0] + T[1,0] + T[0,1] )/4.0
        T_new[0,50] = ( 2.0*T[0,0] + T[1,0] + T[0,1] )/4.0
        T_new[50,50] = ( 2.0*T[0,0] + T[1,0] + T[0,1] )/4.0

        
        T[:,:] = T_new[:,:]
        counter += 1
```
