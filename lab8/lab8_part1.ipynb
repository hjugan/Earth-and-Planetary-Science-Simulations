```python
L = 1
N = 20
x = np.linspace(0, L, N+1)
T = np.zeros(21)
```

```python
for i in range(1, 11):
    T[i] = (0.1 * i)

for i in range(11, 21):
    T[i] = 2 + (-0.1 * i)
print(T)
```

```python
plt.plot(x,T)
```

```python
# Tnew[i] = k * (dt) * (1/(dx**2)) * (T[i+1] - 2*T[i] + T[i-1]) + T[i]
k = 1
dt = 1e-4
dx = .05
Tnew = np.zeros(21)
for i in range(20):
    Tnew[i] = k * (dt) * (1/(dx**2)) * (T[i+1] - 2*T[i] + T[i-1]) + T[i]
T[:] = Tnew[:]
plt.plot(x,T)
```

```python
plt.clf()
plt.plot(x,T, 'ro-',mfc='w')
plt.show()
plt.draw()
plt.pause(0.05) # choose the time argument between 0.01 and 0.5
```

```python
plt.clf()
for i in range(1, 11):
    T[i] = (0.1 * i)

for i in range(11, 21):
    T[i] = 2 + (-0.1 * i)
Tnew = np.zeros(21)
step = 0
initial_temp = np.sum(T)
while True:
    for i in range(1, 20):
        Tnew[i] = k * (dt) * (1/(dx**2)) * (T[i+1] - 2*T[i] + T[i-1]) + T[i]
    T[:] = Tnew[:]
    step += 1
    current_temp = np.sum(T)
    if step % 25 == 0:
        plt.clf()
        plt.ylim(0,1)
        plt.plot(x,T, 'ro-',mfc='w')
        plt.show()
        plt.draw()
        plt.pause(0.05) # choose the time argument between 0.01 and 0.5
    temp_ratio = current_temp / initial_temp
    if temp_ratio < 0.01:
        time = step * dt
        print(f"The sim took {time} seconds")
        break
```

When running until less than 1% of the initial heat had dissipated, the sim took 4706 time steps so 0.4706 seconds

The heat is leaving from the boundaries. At every iteration, the heat is passed from the local maxima, to the local minima. Because the end points have nowhere to pass the heat to, it just disappears. The heat is basically just passed down the line to the end points. It can't move backward from the endpoints because the temperature is higher, and it can't raise the temperature of the end points, because that would move against the flow of the heat.

The curve gets smoother with more iterations because the rate of change of the temperature is proportional to the size of the local maximum within the data set. When the simulation starts, the local maximum is humongous compared to the end points. This results its a steeper curve, but as heat flows away from it, the peak is closer to its neighboring values and it has a slower diffusion rate. This makes the curve smoother and spread out slower.

```python
plt.clf()
k = 1
dx = 0.05
dt = 1.3e-3
T = np.zeros(21)
Tnew = np.zeros(21)
for i in range(1, 11):
    T[i] = (0.1 * i)

for i in range(11, 21):
    T[i] = 2 + (-0.1 * i)
initial_temp = np.sum(T)
step = 0
while True:
    for i in range(1, 20):
        Tnew[i] = k * (dt) * (1/(dx**2)) * (T[i+1] - 2*T[i] + T[i-1]) + T[i]
    T[:] = Tnew[:]
    step += 1
    current_temp = np.sum(T)
    if step % 5 == 0:
        plt.clf()
        plt.ylim(0,1)
        plt.plot(x,T, 'ro-',mfc='w')
        plt.show()
        plt.draw()
        plt.pause(0.05) # choose the time argument between 0.01 and 0.5
    temp_ratio = current_temp / initial_temp
    if temp_ratio < 0.01:
        print(temp_ratio)
        time = step * dt
        print(f"The sim took {time} seconds")
        break
```

The critical time step is at ~1.3e-3 seconds. 

The reason this disrupts the simulation is because having a large dt can cause the heat to spread upwards since you are multiplying the neighboring points by a high value, which may end up being greater than the initial heat that is spreading.
