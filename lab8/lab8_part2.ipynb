```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import PillowWriter
%matplotlib osx
```

```python
n = 21

f = np.zeros(n)
for i in range(1, 11):
    f[i] = 0.1 * i
for i in range(11, 21):
    f[i] = 2 - 0.1 * i


T = np.zeros((n,n))
for i in range(n):
    for j in range(n):
        T[i, j] = min(f[i], f[j]) * 100
# 3D surface, like the one in the PDF
X, Y = np.meshgrid(np.arange(n), np.arange(n))
ax = plt.figure().add_subplot(projection='3d')
ax.plot_surface(X, Y, T, cmap='jet')
```

```python
plt.clf()
plt.imshow(T, origin='lower')
plt.colorbar()
```

```python
dt = 0.05
dy = dx = 0.05
k = 1
dt = 1e-4
eta = k*dt/dx**2
```

```python
fig = plt.figure()
writer = PillowWriter(fps=15)

Tnew = np.zeros((n,n))
initial_temp = np.sum(T)
counter = 0

with writer.saving(fig, "heat_pyramid_2d.gif", dpi=80):
    while True:
        for j in range(1, 20):
            for i in range(1,20):
                Tnew[i,j] = eta * (T[i+1,j] + T[i-1,j] + T[i, j+1] + T[i, j-1] - 4*T[i,j]) + T[i,j]
    
        T[:] = Tnew[:]
        counter += 1
        current_temp = np.sum(T)
        temp_ratio = current_temp / initial_temp

        if counter % 25 == 0:
            fig.clf()
            ax = fig.add_subplot(projection='3d')
            ax.plot_surface(X, Y, T, cmap='jet', vmin=0, vmax=100)
            ax.set_zlim(0, 100)
            ax.set_title(f"t = {counter*dt:.4f}")
            writer.grab_frame()
        
        if temp_ratio < 0.01:
            break
```
